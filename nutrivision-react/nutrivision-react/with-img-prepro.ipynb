```python
model_path = "/kaggle/input/models/deepikajetti/final-model/other/default/1/final_model"

# ✅ SAFE processor loading
try:
    from transformers import AutoProcessor
    processor = AutoProcessor.from_pretrained(
        model_path,
        local_files_only=True,
        trust_remote_code=True
    )
except:
    print("⚠️ Falling back to default SigLIP processor")
    processor = AutoImageProcessor.from_pretrained(
        "google/siglip-base-patch16-224"
    )

# Load config
with open(os.path.join(model_path, "config.json")) as f:
    old_config = json.load(f)

id2label = old_config["id2label"]
label2id = old_config["label2id"]

print("✅ Total classes:", len(label2id))
```

```python
dataset1_path = "/kaggle/input/datasets/iamsouravbanerjee/indian-food-images-dataset/Indian Food Images/Indian Food Images"
dataset2_path = "/kaggle/input/datasets/deepikajetti/food-dataset-custom/food_dataset"

dataset_labels = set()

# dataset1
for cls in os.listdir(dataset1_path):
    if os.path.isdir(os.path.join(dataset1_path, cls)):
        dataset_labels.add(cls.lower().replace(" ", "_"))

# dataset2
for cls in os.listdir(dataset2_path):
    if os.path.isdir(os.path.join(dataset2_path, cls)):
        dataset_labels.add(cls)

model_labels = set(label2id.keys())

missing = model_labels - dataset_labels

print("❌ Extra class in model:", missing)
```

```python
old_dataset_path = "/kaggle/input/datasets/iamsouravbanerjee/indian-food-images-dataset/Indian Food Images/Indian Food Images"

image_paths = []
labels = []

def add_dataset(base_path, transform_name=False):
    for cls in os.listdir(base_path):

        cls_path = os.path.join(base_path, cls)
        if not os.path.isdir(cls_path):
            continue

        cls_name = cls.lower().replace(" ", "_") if transform_name else cls

        if cls_name not in label2id:
            continue

        for img in os.listdir(cls_path):
            if img.lower().endswith((".jpg", ".jpeg", ".png", ".webp", ".jfif", ".avif")):
                image_paths.append(os.path.join(cls_path, img))
                labels.append(label2id[cls_name])

# Load both datasets
add_dataset(dataset_path)
add_dataset(old_dataset_path, transform_name=True)

print("✅ Total images:", len(image_paths))
print("✅ Unique classes used:", len(set(labels)))

dataset = Dataset.from_dict({
    "image": image_paths,
    "label": labels
})
```

```python
print(min(labels), max(labels))
```

```python
image_size = processor.size.get("height", processor.size.get("shortest_edge", 224))

train_transforms = T.Compose([
    T.RandomResizedCrop(image_size, scale=(0.6, 1.0)),
    T.RandomHorizontalFlip(),
    T.RandomVerticalFlip(p=0.2),

    T.ColorJitter(
        brightness=0.4,
        contrast=0.4,
        saturation=0.4,
        hue=0.1
    ),

    T.RandomRotation(25),
    T.RandomGrayscale(p=0.1),

    T.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0)),

    T.ToTensor(),

    T.Normalize(mean=processor.image_mean, std=processor.image_std)
])

eval_transforms = T.Compose([
    T.Resize((image_size, image_size)),
    T.ToTensor(),
    T.Normalize(mean=processor.image_mean, std=processor.image_std)
])
```

```python
error_count = 0

def preprocess_train(example):
    global error_count
    images = []

    for path in example["image"]:
        try:
            img = Image.open(path).convert("RGB")
            img = train_transforms(img)
            images.append(img)
        except Exception as e:
            error_count += 1
            print(f"⚠️ Error loading: {path}")
            images.append(torch.zeros(3, image_size, image_size))

    return {"pixel_values": images}


def preprocess_eval(example):
    images = []

    for path in example["image"]:
        try:
            img = Image.open(path).convert("RGB")
            img = eval_transforms(img)
            images.append(img)
        except Exception as e:
            images.append(torch.zeros(3, image_size, image_size))

    return {"pixel_values": images}
```

```python
from datasets import ClassLabel

dataset = dataset.cast_column(
    "label",
    ClassLabel(num_classes=len(label2id))
)

dataset = dataset.train_test_split(
    test_size=0.1,
    seed=42,
    stratify_by_column="label"
)

train_dataset = dataset["train"]
eval_dataset = dataset["test"]
```

```python
print("Train size:", len(train_dataset))
print("Eval size:", len(eval_dataset))
```

```python
train_dataset = train_dataset.map(
    preprocess_train,
    batched=True,
    batch_size=16,   # ✅ smaller
    remove_columns=["image"]
)

eval_dataset = eval_dataset.map(
    preprocess_eval,
    batched=True,
    batch_size=8,
    remove_columns=["image"]
)

# Rename
train_dataset = train_dataset.rename_column("label", "labels")
eval_dataset = eval_dataset.rename_column("label", "labels")

train_dataset.set_format(type="torch", columns=["pixel_values", "labels"])
eval_dataset.set_format(type="torch", columns=["pixel_values", "labels"])
```

```python
model = SiglipForImageClassification.from_pretrained(
    model_path,
    num_labels=len(id2label),
    id2label=id2label,
    label2id=label2id,
    ignore_mismatched_sizes=True,
    local_files_only=True   # ✅ IMPORTANT
)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
```

```python
accuracy = evaluate.load("accuracy")

def compute_metrics(eval_pred):
    logits, labels = eval_pred

    # Handle tuple case (sometimes logits wrapped)
    if isinstance(logits, tuple):
        logits = logits[0]

    preds = np.argmax(logits, axis=1)
    return accuracy.compute(predictions=preds, references=labels)
```

```python
def collate_fn(batch):
    return {
        "pixel_values": torch.stack([x["pixel_values"] for x in batch]),
        "labels": torch.tensor([x["labels"] for x in batch])  # ✅ FIX
    }
```

```python
training_args = TrainingArguments(
    output_dir="/kaggle/working/results",

    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    do_train=True,
    do_eval=True,

    num_train_epochs=10,
    learning_rate=3e-5,

    save_strategy="steps",
    eval_strategy="steps",   # ✅ FIXED

    save_steps=500,
    eval_steps=500,
    logging_steps=100,

    save_total_limit=2,

    remove_unused_columns=False,
    fp16=True,

    load_best_model_at_end=True,

    metric_for_best_model="accuracy",
    greater_is_better=True,
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    data_collator=collate_fn,
    compute_metrics=compute_metrics
)
```

```python
trainer.train()
```

```python
metrics = trainer.evaluate()
print(metrics)
```

```python
save_path = "/kaggle/working/final_food_model"

trainer.save_model(save_path)
processor.save_pretrained(save_path)

print("✅ Model saved at:", save_path)
```

```python
import os
import torch
import numpy as np
from PIL import Image
from tqdm import tqdm

from transformers import AutoImageProcessor, SiglipForImageClassification
from sklearn.metrics import accuracy_score

# PATH
model_path = "/kaggle/working/final_food_model"

# LOAD
processor = AutoImageProcessor.from_pretrained(
    model_path,
    local_files_only=True
)

model = SiglipForImageClassification.from_pretrained(
    model_path,
    local_files_only=True
)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
model.eval()

torch.set_grad_enabled(False)

print("✅ Model loaded on:", device)
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm

all_preds = []
all_labels = []

for sample in tqdm(eval_dataset):
    img = sample["pixel_values"].unsqueeze(0).to(device)
    label = sample["labels"]

    outputs = model(pixel_values=img)
    pred = torch.argmax(outputs.logits, dim=1).item()

    all_preds.append(pred)
    all_labels.append(label)

print("✅ Predictions collected")
```

```python
from sklearn.metrics import accuracy_score

acc = accuracy_score(all_labels, all_preds)
print("✅ Eval Accuracy (manual):", acc)
```

```python
cm = confusion_matrix(all_labels, all_preds)

print("Confusion matrix shape:", cm.shape)
```

```python
# Get misclassified indices
errors = []

for i in range(len(all_labels)):
    if all_labels[i] != all_preds[i]:
        errors.append((all_labels[i], all_preds[i]))

# Count confusion pairs
from collections import Counter
confusion_counts = Counter(errors)

top_confusions = confusion_counts.most_common(20)

print("🔥 Top confused class pairs:")
for (true, pred), count in top_confusions:
    print(f"{true} → {pred} : {count}")
```

```python
# Get top 20 classes involved in confusion
top_classes = list(set([c[0][0] for c in top_confusions] + [c[0][1] for c in top_confusions]))

# Filter predictions
filtered_labels = []
filtered_preds = []

for t, p in zip(all_labels, all_preds):
    if t in top_classes and p in top_classes:
        filtered_labels.append(t)
        filtered_preds.append(p)

cm_small = confusion_matrix(filtered_labels, filtered_preds, labels=top_classes)

plt.figure(figsize=(12,10))
sns.heatmap(cm_small, annot=True, fmt="d", cmap="Blues")
plt.title("Top Confused Classes")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()
```

```python
id2label = model.config.id2label

print("\n🔥 Top Confusions (with names):")

for (true, pred), count in top_confusions:
    print(f"{id2label[int(true)]} → {id2label[int(pred)]} : {count}")
```

# Testing with random images from dataset

```python
import os
import torch
import random
from PIL import Image
from tqdm import tqdm

from transformers import AutoImageProcessor, SiglipForImageClassification
```

```python
model_path = "/kaggle/input/models/deepikajetti/final-food-model/other/default/1"

processor = AutoImageProcessor.from_pretrained(model_path, local_files_only=True)
model = SiglipForImageClassification.from_pretrained(model_path, local_files_only=True)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
model.eval()

id2label = model.config.id2label

print("✅ Model loaded")
```

```python
dataset_path = "/kaggle/input/datasets/deepikajetti/food-dataset-custom/food_dataset"
```

```python
correct = 0
total = 0

results = []

with torch.no_grad():
    for cls in tqdm(sorted(os.listdir(dataset_path))):

        cls_path = os.path.join(dataset_path, cls)
        if not os.path.isdir(cls_path):
            continue

        # Get images
        images = [
            img for img in os.listdir(cls_path)
            if img.lower().endswith((".jpg", ".jpeg", ".png", ".webp", ".jfif", ".avif"))
        ]

        if len(images) == 0:
            continue

        # Pick random image
        img_name = random.choice(images)
        img_path = os.path.join(cls_path, img_name)

        try:
            image = Image.open(img_path).convert("RGB")

            inputs = processor(images=image, return_tensors="pt").to(device)

            outputs = model(**inputs)
            pred_id = torch.argmax(outputs.logits, dim=1).item()

            pred_label = id2label[pred_id]

            total += 1

            if pred_label == cls:
                correct += 1
                status = "✅"
            else:
                status = "❌"

            results.append((cls, pred_label, status))

        except Exception as e:
            print(f"⚠️ Error: {img_path}")
```

```python
for true, pred, status in results:
    print(f"{status} {true} → {pred}")
```

```python
accuracy = correct / total

print("\n🎯 Class-wise Test Accuracy (1 image per class):")
print(f"Accuracy: {accuracy:.4f} ({correct}/{total})")
```

# Testing with multiple images { new dataset }

```python
import os
import torch
import random
from PIL import Image
from tqdm import tqdm

from transformers import AutoImageProcessor, SiglipForImageClassification
```

```python
model_path = "/kaggle/input/models/deepikajetti/final-food-model/other/default/1"

processor = AutoImageProcessor.from_pretrained(model_path, local_files_only=True)
model = SiglipForImageClassification.from_pretrained(model_path, local_files_only=True)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
model.eval()

id2label = model.config.id2label

print("✅ Model loaded")
```

```python
SAMPLES_PER_CLASS = 20

class_correct = {}
class_total = {}

overall_correct = 0
overall_total = 0

with torch.no_grad():
    for cls in tqdm(sorted(os.listdir(dataset_path))):

        cls_path = os.path.join(dataset_path, cls)
        if not os.path.isdir(cls_path):
            continue

        images = [
            img for img in os.listdir(cls_path)
            if img.lower().endswith((".jpg", ".jpeg", ".png", ".webp", ".jfif", ".avif"))
        ]

        if len(images) == 0:
            continue

        # Sample multiple images
        sampled_images = random.sample(images, min(SAMPLES_PER_CLASS, len(images)))

        correct = 0
        total = 0

        for img_name in sampled_images:
            img_path = os.path.join(cls_path, img_name)

            try:
                image = Image.open(img_path).convert("RGB")

                inputs = processor(images=image, return_tensors="pt").to(device)

                outputs = model(**inputs)
                pred_id = torch.argmax(outputs.logits, dim=1).item()

                pred_label = id2label[pred_id]

                total += 1
                overall_total += 1

                if pred_label == cls:
                    correct += 1
                    overall_correct += 1

            except:
                continue

        class_correct[cls] = correct
        class_total[cls] = total
```

```python
overall_acc = overall_correct / overall_total

print("🎯 Overall Accuracy (multi-sample):")
print(f"{overall_acc:.4f} ({overall_correct}/{overall_total})")
```

```python
print("\n📊 Per-Class Accuracy:\n")

for cls in sorted(class_total.keys()):
    acc = class_correct[cls] / class_total[cls]
    print(f"{cls}: {acc:.2f} ({class_correct[cls]}/{class_total[cls]})")
```

```python
print("\n❌ Weak Classes (accuracy < 50%):\n")

for cls in class_total:
    acc = class_correct[cls] / class_total[cls]
    if acc < 0.5:
        print(f"{cls}: {acc:.2f}")
```

```python
print("\n✅ Strong Classes (accuracy ≥ 80%):\n")

for cls in class_total:
    acc = class_correct[cls] / class_total[cls]
    if acc >= 0.8:
        print(f"{cls}: {acc:.2f}")
```
